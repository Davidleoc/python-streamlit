```python
# Passo a passo do projeto
# Passo 1: Abrir a base de dados (Importar a base de dados)

import pandas as pd 

tabela = pd.read_csv("cancelamentos.csv")

# Passo 2: Visualizar a base de dados, e enteder informações disponíveis
    # Identificar possíveis problemas/erros com o dados

tabela = tabela.drop(columns="CustomerID") # Removendo a coluna "CustomerID" da tabela, pois não é necessária para o processo de automação

display(tabela) # Para mostrar em formato de tabela em arquivos .ipynb
```

```python
# Passo 3: Tratar os dados e corrigir problemas
display(tabela.info()) # Para mostrar informações sobre a tabela, como tipos de dados e valores nulos

tabela = tabela.dropna() # Removendo linhas com valores nulos

display(tabela.info()) 

# NaN - not a number, ou seja, valores nulos.
```

```python
# Passo 4: Análise Inicial

display(tabela["cancelou"].value_counts(normalize=True)) # Contagem de valores na coluna "cancelou". Normalize = True, para ver em porcentagem

```

```python
# Passo 5: Análise detalhada
import plotly.express as px

for coluna in tabela.columns:
    if coluna != "cancelou":
        grafico = px.histogram(tabela, x=coluna, color="cancelou", text_auto=True) # As cores do gráfico serão definidas de acordo com a coluna "cancelou"
        grafico.show()
```
